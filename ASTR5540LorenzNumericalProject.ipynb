```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

#Help for solving the IVP came from https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html$0

#Define Lorenz system
def lorenz (t, w, sigma, r, b):
  x, y, z = w
  return [sigma * (y - x), r * x - y - x * z, x * y - b * z]

#Solve IVP using sigma = 3, b = 1 and r = 17
# Initial conditions are [x(0), y(0)], z(0)] = [0, 1, 0]
sol = solve_ivp(lorenz, [0, 50], [0, 1, 0], args=(3, 17, 1), dense_output=True)

#Plot dense solution
t = np.linspace(0, 50, 1000)
w = sol.sol(t)
x, y, z = w

# Create three separate subplots
fig, axs = plt.subplots(3, 1, figsize=(8, 10), sharex=True)

# Plot x(t)
axs[0].plot(t, x, color='r')
axs[0].axhline(-4, color='k', linestyle='--', label="Stable critical point")
axs[0].set_ylabel('x(t)')
axs[0].set_title('Lorenz System')
axs[0].legend()
axs[0].grid(True)

# Plot y(t)
axs[1].plot(t, y, color='g')
axs[1].axhline(-4, color='k', linestyle='--')
axs[1].set_ylabel('y(t)')
axs[1].grid(True)

# Plot z(t)
axs[2].plot(t, z, color='b')
axs[2].axhline(16, color='k', linestyle='--')
axs[2].set_ylabel('z(t)')
axs[2].set_xlabel('t')
axs[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

#Help for solving the IVP came from https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html$0

def lorenz(t, w, sigma, r, b):
    x, y, z = w
    return [sigma*(y - x), r*x - y - x*z, x*y - b*z]

sigma, b = 10, 8/3
r = 13.926

#Define initial value for homoclinic orbit close to origin (help from Claude)

sol = solve_ivp(lorenz, [0, 4], [1e-6,1e-6,1e-6], args=(sigma, r, b),
                rtol=1e-12, atol=1e-14, dense_output=True)

t = np.linspace(0, sol.t[-1], 5000)
x, y, z = sol.sol(t)

# Create three separate subplots
fig, axs = plt.subplots(3, 1, figsize=(8, 10), sharex=True)

# Plot x(t)
axs[0].plot(t, x, color='r')
axs[0].set_ylabel('x(t)')
axs[0].set_title('Lorenz System')
axs[0].grid(True)

# Plot y(t)
axs[1].plot(t, y, color='g')
axs[1].set_ylabel('y(t)')
axs[1].grid(True)

# Plot z(t)
axs[2].plot(t, z, color='b')
axs[2].set_ylabel('z(t)')
axs[2].set_xlabel('t')
axs[2].grid(True)

plt.tight_layout()
plt.show()
```
